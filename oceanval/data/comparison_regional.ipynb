```python

# step 1 is to figure out the variables available
try:
    variables = []
    regions = []
    for dir_name in list(model_dict.values()):
        paths = glob.glob(dir_name + "/oceanval_results/regionals/*")
        region = os.path.basename([x for x in paths if x.endswith(".txt")][0]).replace(".txt", "")
        regions.append(region)
        paths = [x for x in paths if not x.endswith(".txt")]
        variables += [os.path.basename(x).split("_")[1] for x in paths]

    regions = list(set(regions))
    regional_summaries = True
    if len(regions) == 0:
        md("There are no regional summaries")
        regional_summaries = False

    if len(regions) > 1:
        md("Validations did not use consistent subregions, so they will not be compared")
        regional_summaries = False

    # restrict variables to those that occur twice
    if regional_summaries:
        variables = list(set([x for x in variables if variables.count(x) >= 2]))# step 1 is to figure out the variables available
    else:
        variables = []
except:
    md("There do not appear to be any regional summaries available for comparison.")
```

```python
try:
    lon_name = [x for x in ds_regions.to_xarray().coords if "lon" in x][0]
    lat_name = [x for x in ds_regions.to_xarray().coords if "lat" in x][0]
    df_mapped = (
        ds_regions
        .to_dataframe()
        .reset_index()
        # rename the columns
        .rename(columns = {lon_name: "lon", lat_name: "lat"})
        .melt(id_vars = ["lon", "lat"])
        .dropna()
        .merge(regions_contents.loc[:,["variable", "long_name"]])
        .drop(columns = [ "value"])
    )

    def fix_name(x):
        x = x.replace("North East", "NE")
        x = x.replace("North ", "N ")
        if x == "Channel":
            x = "English Channel"
        return x

    fix_name = np.vectorize(fix_name)


    df_mapped.long_name = fix_name(df_mapped.long_name)
    regional = True
except:
    regional = False



```

```python
if regional:
    from matplotlib.colors import ListedColormap
    # one panel per region: Full Domain first, then the order in the regions file
    region_titles = df_mapped.loc[:, ["long_name", "variable"]].drop_duplicates()
    region_titles = pd.concat([
        region_titles.query("long_name == 'Full Domain'"),
        region_titles.query("long_name != 'Full Domain'"),
    ])
    n_regions = len(region_titles)
    n_cols = 3 if n_regions == 3 else int(np.ceil(np.sqrt(n_regions)))
    n_rows = int(np.ceil(n_regions / n_cols))
    nc.panel_plot(
        {str(x): ds_regions for x in region_titles.long_name},
        var = list(region_titles.variable),
        ncol = n_cols,
        nrow = n_rows,
        size = [10, 2.4 * n_rows],
        colours = ListedColormap(["#333333"]),
        limits = [0, 1],
        legend_position = None,
        land = "lightgrey",
        grid_labels = False,
    )
```
